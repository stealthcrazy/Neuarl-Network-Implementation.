```python
hLc = 2 # hidden layer count
hls = 16 # nodes per hidden layer
inps = 784 # nodes in input layer
outs = 10 # outputs 

```

```python
h1w = np.random.normal(0.0,0.5,(inps,hls))
h2w = np.random.normal(0.0,0.5,(hls,hls))
o1w = np.random.normal(0.0,0.5,(hls,outs))

# setting up the weights for each layer ^
#size of each weight is row size = previous layer node size , col size = current layer node size
#using normal dist to intialise weights

h1b = np.random.normal(0.0,0.5,(hls))
h2b = np.random.normal(0.0,0.5,(hls))
o1b = np.random.normal(0.0,0.5,(outs))
#setting up the biases ^
```

```python
# defining the sigmoid

def sig(M):
    return 1/(np.exp(-M)+1)
```

```python
# calculating the sum of squared residuals

def SSR(o , lbl):
    x = np.zeros(10)
    x[lbl] = 1
    
    return np.linalg.norm(x-o)
```
